# Batch Image Explanation

Process multiple images at once using `SFL_batch` and the `explainer.batch()` helper.

```python
import torch
import torch.nn as nn
import torchvision.models as models

from causal_explainer import get_device
from causal_explainer.explainer import batch
```

```python
# Load model
device = get_device()
print(f'Using device: {device}')

model = models.resnet50(pretrained=True)
model = nn.Sequential(model, nn.Softmax(dim=1))
model = model.eval().to(device)
for p in model.parameters():
    p.requires_grad = False
```

```python
# Process all images in data_demo/
results = batch(
    model,
    image_folder='../data_demo/',
    N=100,
    s=8,
    p1=0.2,
    save_maps=False,  # Set True to save to disk instead of displaying
    device=device
)
```

```python
# Inspect results
print(f"Processed {len(results['target_list'])} images")
print(f"Target classes: {results['target_list']}")
print(f"Ochiai arrays shape: {results['ochiai_array'][0].shape}")
```
